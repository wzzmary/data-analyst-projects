```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


data = pd.read_csv("PSM2003.csv", sep=";", encoding='utf-8')

# Données
x = data["region"]  # Colonne région
y = data["idispy"]  # Colonne revenu

# Je fais la moyenne des revenus par région
revenu_par_region = data.groupby("region")["idispy"].mean()

# Moon graphique en barre 
plt.bar(revenu_par_region.index, revenu_par_region.values, color="indigo", alpha=0.6)

#configurons le graphique 
plt.title(" Nombre de revenu moyen selon les régions", fontsize=16)
plt.xlabel("Régions", fontsize=12)
plt.ylabel("Revenu moyen par personne (en ⁄CHF)", fontsize=12)

# La grille pour me permettre de bien voir les données sur l'axe des Y
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()



# Notre graphique montre la distribution des revenus moyens par région. 
# Chaque barre représente la moyenne des revenus pour une région donnée.
# L'objectif est d'identifier des disparités ou des tendances régionales dans les revenus. 
# Par exemple, certaines régions pourraient avoir des revenus moyens significativement plus élevés ou plus bas que d'autres. 
# Cette analyse peut aider à déterminer s'il existe des inégalités économiques entre les régions ou à cibler des interventions spécifiques
```

```python


data = pd.read_csv("PSM2003.csv", sep=";", encoding="utf-8")

# Tous les niveaux d'éducation 
education_levels = ["PhD", "apprenti", "bachelor", "universi", "vocation", " compulso"," domestic "
,"elementa"," full-tim", "general", "incomple"," technica" ]

data_filtered = data[data["edcat"].isin(education_levels)]

# Les niveaux d'éducation par région
education_by_region = data_filtered.groupby(["region", "edcat"]).size().unstack()

education_by_region.plot(kind="bar",
figsize=(12, 6), stacked=True, colormap="rainbow")

#configurons le graphique 
plt.title("Niveaux d'éducation par régions", fontsize=20)
plt.xlabel("Régions", fontsize=16)
plt.ylabel("Nombre de personnes", fontsize=16)
plt.legend(title="Niveau d'éducation", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()


plt.show()


# Ce graphique représente la répartition des niveaux d'éducation par région.
# Chaque barre empilée montre le nombre de personnes ayant différents niveaux d'éducation dans chaque région.
# L'objectif est d'identifier les tendances dans l'éducation régionale.
# EXEMPLE:   Quelles régions ont des niveaux d'éducation plus élevés (par exemple, un grand nombre de PhD ou de bachelors) ?
# EXEMPLE:   Quelles régions ont des niveaux d'éducation plus faibles (par exemple, un grand nombre de "elementa" ou "incomple") ?
# CELA NOUS MONTRE QU'IL PEUT Y AVOIR DES REGIONS AVEC BEAUCOUP DE REVENUS (EX: DANS LE GRAPHIQUE PRECEDANT region de "Central") avec des niveau d'education bas.
```

```python

import pandas as pd
# on importe le module `numpy` sous le nom court `np`
import numpy as np
# on importe le module `pyplot` de `matplotlib` sous le nom court `plt`
import matplotlib.pyplot as plt

data = pd.read_csv("PSM2003.csv", sep=";", 
encoding='utf-8')

# Données
x = data["nbpers"]  
y = data["idispy"] 

# Régression linéaire en utilisant une fonction polyfit() 
#avec un degré 1 prédéfini dans le module numpy. 
#Cette fonction calcule les coefficients de la fonction polynomiale de degré 1 (ligne).
b, a = np.polyfit(x, y, 1)

# Nous transformons x en un vecteur de valeurs, 
#chaque élément de ce vecteur est multiplié par 
#le coefficient b et la valeur de a est ajoutée à chaque élément.
y_pred = b * x + a

#nous imprimons les données en utilisant un graphique à nuage de points.

plt.scatter(x, y, color="blue", alpha=0.6, label="Données")
plt.plot(x, y_pred, color="red", label="Ligne de régression")

#nous configurons le graphique (donnons un titre,
#les noms des axes, la légende)...
plt.title("Revenu en fonction du nombre de personnes par ménage", fontsize= 16)
plt.xlabel("Nombre de personne par menage", fontsize= 12)
plt.ylabel("Revenu (en CHF)",fontsize= 12)
plt.legend()
plt.show()





# Chaque point bleu représente les données réelles, tandis que la ligne rouge correspond
# à une régression linéaire qui modélise la tendance générale des données.
# L'objectif principal de ce graphique est d'observer s'il existe une corrélation
# entre la taille du ménage et le revenu. 
#EXEMPLE: la ligne de régression peut indiquer si un ménage avec plus de membres 
# tend à avoir un revenu plus élevé ou plus faible. Maia nous voyons qu'une majorite de menage de 1,2 voire 3 personnes ont des revenues presque égaux à ceux avec des menages de 5,6 ou 7 personnes  
```

```python

data = pd.read_csv("PSM2003.csv", sep=";", encoding="utf-8")

# Mes variables 
x = data["age"]  
y = data["idispy"]  

b, a = np.polyfit(x, y, 1)  
y_pred = b * x + a 

# nous tracons les données
plt.scatter(x, y, color="blue", alpha=0.5, label="Données réelles")  # Points réels
plt.plot(x, y_pred, color="red", label="Ligne de régression")  # Ligne de régression

# nous configurons le graphique
plt.title("Relation entre l'âge et le revenu total du ménage", fontsize=16)
plt.xlabel("Âge de la personne de référence", fontsize=12)
plt.ylabel("Revenu total ( en CHF)", fontsize=12)
plt.legend()
plt.show()


# Le graphique présent nous montre la relation entre l'âge de la personne de référence et le revenu total du ménage.
# La ligne de régression,  en rouge que je peux voir sur le graphique généré, est plate et ca nous  montre qu'il n'y a  pas de relation  entre l'âge et le revenu.
# Ca nous suggère ou nous montre que l'âge tout seul n'est pas un bon prédicteur des revenus des ménages.
```
